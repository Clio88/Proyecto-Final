# Introducción:

Para empezar con la estadística inferencial se debe de definir lo que es una muestra y consigo mismo que es una población.


> <h3> <font color= "3672ec";>Una población es un conjunto de personas u objetos de los que se desea conocer algo a través de una investigación.</h3>

> <h3> <font color= "3672ec";>Una muestra aleatoria es una colección de variables aleatorias $X_1,\ldots,X_n$ que son independientes e idénticamente distribuidas.</h3>

Algunas caracteristicas de las muestras, es que son:

- Aleatorias
- Representativas
- Tienen estimadores
- Tienen un parámetro desconocido a estimar

Para poder conocer lo que se estudia sobre la población se utilizan las muestras aleatorias. Dichas muestras traen consigo los parámetros que se quieren estudiar y para ello se utilizan las **distribuciones muestrales**.

Sean $Y_1,\ldots,Y_n$ variables aleatorias (v.a) independientes que se distribuyen como una Normal con parámetros $𝜇$ y $σ^2$ $⇒$

$$\bar{Y}=\frac{1}{n}\sum_{i=1}^{n}Y_i\sim N\left(𝜇,\frac{\sigma^2}{n}\right)$$

Ahora si se realizan los siguientes pasos:

$$\begin{align*}Z_i&=\frac{Y_i-𝜇}{\sqrt{σ^2}}\\
\hspace{1cm}\\
&=\frac{Y_i-𝜇}{σ}∼N(0,1)
\end{align*}$$

Dichos pasos se conocen como la estandarización de la observación. Y si se elevan al cuadrado y se realiza la suma de cada estadarización de cada observación, se obtiene una nueva variable aleatoria cuya distribución se da como una **chi-cuadrada con $n$ grados de libertad**

$$\sum_{i=1}^{n}Z^2∼\chi_n^2$$

Esta distribución chi-cuadrada es un pilar fundamental, porque abre la puerta a realizar inferencia sobre la varianza poblacional. Entonces si se quiere realizar una inferencia acerca de la media de una población Normal $(𝜇)$ con $σ^2$ desconocida, se puede estimar.

# #Estimadores

Una estimación siempre buscará el "mejor parámetro" que se adapte y sea bueno para el conjunto de datos. Pero primero se debe de ver la definición de un estimador.


> <h3> <font color= "3672ec";>Un estimador puntual para un parámetro $θ$ es una función de una muestra aleatoria $X_1,\ldots,X_n$ que se usa para estimar $θ$.</h3>

De manea similar se debe de ver la definición de un estadístico.

> <h3> <font color= "3672ec";>El conjunto de v.a. $\left(X_1,\ldots,X_n \right)$ se denomina como una muestra aleatoria de tamaño $n$, cualquier función real de una muestra aleatoria, $T_{(X_1,\ldots,X_n)}(X_1,\ldots,X_n)$ se llama estadístico.</h3>

Y un estimador de $θ$ es cualquier estadístico $T_{(X_1,\ldots,X_n)}(X_1,\ldots,X_n)$ se denota por:

$$\hat{θ}=T(X_1,\ldots,X_n)$$

De manera continua se define la función de densidad de la muestra.


> <h3> <font color= "3672ec";>Sea $X$ una v.a. con función de densidad $f(x;θ)$ que depende de los parámetros desconocidos $θ$. Sea $(X_1, \ldots, X_n)$ una m.a. de $X$ se tiene: $$f(x;θ)=f(x_1,\ldots,x_n)=\prod_{i=1}^nf(x_i;θ)$$</h3>

Otra definición importante es del error cuadrático medio, dicho "error" muestra la diferencia entre el estimador y el valor verdadero de $θ$ y se define de la siguiete forma:

$$\begin{align*}ECM(\hat{θ})&=\mathbb{E}\left[(\hat{θ}-θ)^2\right]\\
\hspace{1cm}\\
&=Var(\hat{θ})+\left(\mathbb{E}\left[\hat{θ}\right]-θ\right)^2\end{align*}$$


## Propiedades de los estimadores

Se tienen dos tipos de estimadores:
- Puntuales, los cuales son en función de un solo valor.
- Intervalos, los cuales son en función en un rango de valores.

Algunas de las propiedades que se buscan en los estimadores es que:
- Sean insesgados
- Sean consistentes
- Sean completos
- Sean suficientes

De esa manera definiremos cada una de las propiedades mencionadas anteriormente.

> <h3> <font color= "3672ec";>Sea $(X_1,\ldots,X_n)$ una m.a. que proviene de una v.a. con $f(x)$ donde $\mathbb{E}(X)=𝜇$ y $Var(X)=σ^2$ $$T(x)=\bar{X}=\frac{X_1+\cdots+X_n}{n}$$ dicho estimador es insesgado si $\mathbb{E}(\bar{X})=𝜇$.</h3>

De esa manera un estimador $\hat{θ}=T(X_1,\ldots,X_n)$ es **insesgado de $θ$** si $\mathbb{E}[\hat{θ}]=θ$.

Y si un estimador $\hat{θ}$ es insesgado, entonces su error cuadratico medio es la varianza.
$$\mathbb{E}\left[(\hat{θ}-θ)^2\right]=Var(θ)$$


> <h3> <font color= "3672ec";>Sea $T(X_1,\ldots,X_n)=\hat{θ}$ un estimador de $θ$, se dice que es consistente si $T$ converge en probabilidad a $\theta$ cuando el tamaño de la muestra tiende a infito, es decir, si: $$lim_{n→∞}\mathbb{P}(|\hat{θ}-θ|<𝛆)=1 \hspace{0.5cm} ∀  𝛆>0$$</h3>

Dicha propiedad garantiza que cualquier estimación se acercará o tendera a $θ$.

> <h3> <font color= "3672ec";>Un estimador $(X_1,X_2,\ldots,X_n)=\hat{θ}$ es completo si $\mathbb{E_{θ}}[g(\hat{θ})]=0$ para todo $θ$, entonces $g(\hat{θ})=0$ casi seguramente.</h3>

> <h3> <font color= "3672ec";>Un estimador $\hat{θ}$ es suficiente cuando $f(x_1,\ldots,x_n|\hat{θ})$ no depende del parametros a estimar $θ$.</h3>

Ahora se definira el sesgo de un estimador $\hat{θ}$.

> <h3> <font color= "3672ec";>El sesgo B de un estimador $\hat{θ}$ se define como: $$B=\mathbb{E}\left[\hat{θ}\right]-θ$$.</h3>

Si $\mathbb{E}\left[\hat{θ}\right]>θ$, entonces el estimador es sesgado positivamente
Si $\mathbb{E}\left[\hat{θ}\right]<θ$, entonces el estimador es sesgado negativamente.

#### Minimización del ECM

Sea $X$ una v.a. con segundo momento de orden finito como:
$$g(u)=\mathbb{E}\left[(X-u)^2\right]$$

Entonces $g(u)$ se minimiza cuando $\mathbb{E}[X]=u$
Por lo tanto $Var(X)\leq \mathbb{E}\left[(X-u)^2\right]$

### #Método de momentos

Para definir el método de momentos se debe de recordar la definición de la Función Generadora de Momentos (FGM).

> <h3> <font color= "3672ec";>La función generadora de momentos de una variable aleatoria discreta o continua $X$ es la función $M(t)$ definida como sigue: $$M(t)=\mathbb{E}(e^{tX})$$ para valores reales de $t$ en donde la esperanza existe.</h3>

La función generadora de momentos en el $n$-ésimo momento evaluado en cero, es decir $\mathbb{E}[X^n]$ se le conoce como los momentos poblacionales mientras que se definen los momentos muestrales como:

$$M_n=\frac{1}{n}\sum_{i=1}^kx_i^n$$

De esa manera, para el método de momentos se deben de igualar los momentos poblacionales con los muestrales

####<font color= "8f2bed"> Ejemplo:</font>
Hallar los estimadores por el método de momentos para $(X_1,\ldots,X_n)$ una m.a. de una población con densidad $Gamma(α,β)$


Si $X∼Gamma(α,β)$, entonces $\mathbb{E}[X]=\frac{α}{β}$ y $Var(X)=\frac{α}{β^2}$

Pero recordemos que:
$$Var(X)=\mathbb{E}[X^2]-(\mathbb{E}[X])^2$$

$⇒$

$$\begin{align*}\mathbb{E}[X^2]&=Var(X)-\left(\mathbb{E}[X]\right)^2\\
\hspace{1cm}\\
&=\frac{α}{β^2}+\frac{α^2}{β^2}\\
\hspace{1cm}\\
&=\frac{α(1+α)}{β^2}\end{align*}$$

Entonces se iguala de la siguiente forma:

$$\begin{align*}\mathbb{E}\left[ X \right]&=\frac{1}{n}\sum X_i\\
\hspace{1cm}\\
\mathbb{E}\left[X^2\right]&=\frac{1}{n}\sum X_i^2
\end{align*}$$

De esa manera la primera sustitución es:

$$\frac{α}{β}=\frac{1}{n}\sum X_i$$

$⇒$

$$\frac{α}{β}=\bar{X} \hspace{1cm} ⇒ \hspace{1cm} \hat{β}=\frac{α}{\bar{X}}$$

Y utilizando $\hat{β}$, se sustituye sobre la segunda igualdad:

$$\frac{\alpha}{β^2}+\frac{α^2}{β^2}=\frac{1}{n}\sum X_i^2$$

Se realizan los despejes y se obtienen los estimadores:

$$\hat{α}=\frac{n\bar{X^2}}{\sum(X_i-\bar{X})^2} \hspace{1cm} \hat{β}=\frac{n\bar{X}}{\sum(X_i-\bar{X})^2}$$




####<font color= "f24421"> Ejercicio:</font>
Sea $(X_1,...,X_n)$ una m.a. con función de densidad:
$$f(x;θ)=θ(1+x)^{-(1+θ)}  \mathbb{I}_{(0,-∞)}(x)$$

**<font color="red">[Estimador de $θ$]</font>**

Sabemos que el primer momento de la función de densidad, $𝜇_1$ es:

$$𝜇_1=\mathbb{E}[X]=∫_{0}^{∞}x⋅f(x)dx=∫_{0}^{∞}x⋅θ(1+x)^{-(1+θ)}dx$$

$$𝜇_1=θ∫_{0}^{∞}x(1+x)^{-(1+θ)}dx$$


Para resolver la integral se hace un cambio de variable. Entonces, sea $u=1+x ⇒ du=dx$ y $x=u-1$

Y de igual manera los límites de la integral cambian, ya que si $x=0 ⇒ u=1$

$⟹$
$$\begin{align*}&𝜇_1=θ∫_{0}^{∞}x(1+x)^{-(1+θ)}dx\\
\hspace{1cm}\\
&𝜇_1=θ∫_{1}^{∞}(u-1)u^{-(1+θ)}du\\
\hspace{1cm}\\
&𝜇_1=θ \left( ∫_{1}^{∞}u^{-θ}du+∫_{1}^{∞}u^{-(1+θ)}du \right)\\
\hspace{1cm}\\
&𝜇_1=θ\left(\frac{1}{θ-1} - \frac{1}{θ}\right)\\
\hspace{1cm}\\
&𝜇_1=θ\left(\frac{θ-(θ-1)}{θ(θ-1)}\right)\end{align*}$$

$⟹$
$$𝜇_1=\frac{1}{θ-1}$$ con $θ>1$

Ahora procedemos a obtener el estimador, sabiendo que $𝜇_1=m_1$, con:

$$m_1=\bar{X}=\frac{1}{n}∑_{i=1}^{n}X_i$$

$⟹$

$$\frac{1}{θ-1}=\bar{X}$$

Despejamos a $θ$ y obtenemos:

$$\hat{θ}=1+\frac{1}{\bar{X}}$$


**<font color="red">[Función Generadora de Momentos (FGM)]</font>**

Recoredemos que la función generadora de momentos se define como:

$$M_x(t)=\mathbb{E}[e^{tX}]$$

Por definición se tiene:

$$M_x(t)=∫_{0}^{∞}e^{tx}θ(1+x)^{-(1+θ)}dx$$

Realizamos un cambio de variable $u=x+1 ⇒ du=dx$. Y de igual manera, cuando $x=0 ⇒ u=1$.

$$M_x(t)=θe^{-t}∫_{1}^{∞}e^{tu}u^{-(1+θ)}du$$

Y se calcula el primer momento como:

$$\begin{align*}&M'_x(t)=\frac{\partial M_x(t)}{\partial t}\\
\hspace{1cm}\\
&M'_x(t)=\frac{\partial}{\partial t}\left(θe^{-t}∫_{1}^{∞}u^{-1+θ}du\right)\\
\hspace{1cm}\\
&M'_x(t)=θe^{-t}\left(-∫_{1}^{∞}e^{tu}u^{-(1+θ)}du + ∫_{1}^{∞}e^{tu}u^{-θ}du\right)\end{align*}$$

Sabemos que:

$$M_x(t)=∫_{0}^{∞}e^{tx}θ(1+x)^{-(1+θ)}dx$$

$⟹$

$$M'_x(t)=-M_x(t)+θe^{-t}∫_{1}^{∞}e^{tu}u^{-θ}du$$

Evaluamos con $t=0$ para obtener $E[X]$. Sabemos que cuando $t=0 ⇒ M_x(t)=1$, por propiedad de la FGM y $M'_x(t)=E[X]$.

Entonces, obtenemos:

$$\begin{align*}&\mathbb{E}[X]=-1+θ∫_{1}^{∞}u^{θ}\\
\hspace{1cm}\\
&\mathbb{E}[X]=\frac{1}{θ-1}\end{align*}$$


Ahora calculamos el segundo momento. Por definición tenemos:

$$M''_x(t)=\mathbb{E}[X^2]=∫_{0}^{∞}x^2⋅θ(1+x)^{-(1+θ)}dx$$

$\hspace{1cm}$

Hacemos un cambio de variable con: $u=1+x ⇒ du=dx$ y $x=u-1$. Los límites cambian y obtenemos:

$\hspace{1cm}$

$$\begin{align*}&M''_x(t)=θ∫_{1}^{∞}(u-2)^2 u^{-(1+θ)}du\\
\hspace{1cm}\\
&M''_x(t)=θ∫_{1}^{∞}\left(u^2⋅u^{-(1+θ)}-2u⋅u^{-(1+θ)}+u^{-(1+θ)}\right)du\\
\hspace{1cm}\\
&M''_x(t)=θ∫_{1}^{∞}\left(u^{1-θ}-2u^{-θ}+u^{-(1+θ)}\right)du\end{align*}$$

Resolvemos cada integral por aparte, recordando que podemos "separar" la integral.

$⟹$

$$\begin{align*}&∫_{1}^{∞}u^{(1-θ)}du=\frac{1}{θ-2}, \hspace{0.2cm} \text{con} \hspace{0.2cm} θ>2\\
\hspace{1cm}\\
&∫_{1}^{∞}u^{-θ}du=\frac{1}{θ-1}, \hspace{0.2cm} \text{con} \hspace{0.2cm} θ>1\\
\hspace{1cm}\\
&∫_{1}^{∞}u^{-(1+θ)}du=\frac{1}{θ}\end{align*}$$

Sustituimos las integrales y obtenemos el segundo momento:

$$M''_x(t)=θ\left(\frac{1}{θ-2}-\frac{1}{θ-1}+\frac{1}{θ}\right), \hspace{0.2cm} \text{con} \hspace{0.2cm} θ>2$$

### #Método de máxima verosimilitud

Para el uso del método de máxima verosimilitud se debe de definir la función de verosimilitud.

> <h3> <font color= "3672ec";>La función de verosimilitud de una muestra aleatoria $X_i,\ldots,X_n$, denotada por $L(θ)$, se define como la función de densidad conjunta: $$L(θ)=f_{X_1,\ldots,X_n}(x_1,\ldots,x_n;θ)$$.</h3>

De esa manera podemos ver la siguiente igualdad:
$$L(θ)=\prod_{i=1}^nf(x_i;θ)$$

Para dicho método también se utiliza la función logarítmica de la verosimilitud:

$$log(L(θ))=\sum_{i=1}^n log(f(x_i;θ))$$

El método de máxima verosimilitud permite obtener el valor de $θ$ que maximiza $L(θ)$, de esa manera se le llama EMV.

Este método proporciona al estimador la propiedad de la invarianza la cual se puede definir de la siguiente forma:

> <h3> <font color= "3672ec";>Si $\hat{θ}$ es EMV de $θ$, entonces si se busca estimar una función de $θ$, digamos $g(θ) ⇒$ $$g(\hat{θ}) \text{ es el EMV de } g(θ)$$</h3>

Por ejemplo si $\hat{θ}=\bar{X}$ es el EMV de $θ ⇒$ $g(θ)=θ^2 ⇒ \hat{g(θ)}=(\bar{X})^2$

####<font color= "8f2bed"> Ejemplo:</font>
Sea $X_1,\ldots,X_n$ una m.a. $Poisson(λ)$, entonces:
- Encontrar el EMV
- ¿$\hat{θ}$ es insesgado?
- Calcular el ECM.



**<font color="red">[Encontrar el EMV]</font>**

Se tiene:

$$L(x_1,\ldots,x_n;θ)=\prod_{i=1}^ne^{-θ}\frac{θ^{x_i}}{x_i!}=\frac{e^{-nθ}θ^{\sum x_i}}{x_1!\cdots x_n!}$$

$⇒$

$$\begin{align*}ln(L(x_1,\ldots,x_n;θ))&=ln\left(\frac{e^{-nθ}θ^{\sum x_i}}{x_1!\cdots x_n!}\right)\\
\hspace{1cm}\\
&=ln(e^{-nθ})-ln(x_1!⋯x_n!)\\
\hspace{1cm}\\
&=-nθ+\sum x_i ln(θ)-\sum ln(x_i!)\end{align*}$$

$⇒$

$$\frac{\partial ln(L(x_1,\ldots,x_n;θ))}{∂θ}=-n+\sum x_i\frac{1}{θ}$$

$\text{Se iguala a 0 la derivada }⇒$

$$\begin{align*}-n+\sum x_i\frac{1}{θ}&=0\\
\hspace{1cm}\\
\frac{1}{θ}\sum x_i&=n\\
\hspace{1cm}\\
\frac{\sum x_i}{n}&=\theta\\
\hspace{1cm}\\
\hat{θ}&=\bar{X}\end{align*}$$

**<font color="red">[¿$\hat{θ}$ es insesgado?]</font>**

$$\begin{align*}\mathbb{E}\left[\hat{θ}\right]&=\mathbb{E}\left[\frac{\sum X_i}{n}\right]\\
\hspace{1cm}\\
&=\frac{1}{n}\mathbb{E}\left[\sum X_i\right]\\
\hspace{1cm}\\
&=\frac{\sum θ}{n}\\
\hspace{1cm}\\
&=\frac{nθ}{n}=θ
\end{align*}$$

**<font color="red">[Calcular el ECM]</font>**

$$\begin{align*}ECM(\hat{θ})&=Var(\hat{θ})\\
\hspace{1cm}\\
&=Var\left(\frac{\sum X_i}{n}\right)\\
\hspace{1cm}\\
&=\frac{Var(\sum X_i)}{n^2}\\
\hspace{1cm}\\
&=\frac{\sum θ}{n^2}\\
\hspace{1cm}\\
&=\frac{nθ}{n^2}=\frac{θ}{n}
\end{align*}$$

####<font color= "f24421"> Ejercicio:</font>
Sea una muestra aleatoria $X_1, \ldots, X_n \overset{iid}{\sim} \mathcal{N}(\mu, \sigma^2) $.
Queremos encontrar el estimador de máxima verosimilitud para $\mu$ y $\sigma^2$.

La función de densidad es:
$$ f(x; \mu, \sigma^2) = \frac{1}{\sqrt{2\pi \sigma^2}} \exp\left(-\frac{(x - \mu)^2}{2\sigma^2}\right) \cdot \mathbf{1}_{(-\infty, \infty)}(x)$$

**<font color="red">[Obtener los EMV para $𝜇$ y $σ^2$]</font>**

Por definición la función de verosimilitud es:

$$\begin{align*}L(𝜇,σ^2)&=∏_{i=1}^{n}f(x_i;𝜇,σ^2)\\
&\hspace{1cm}\\
&=∏_{i=1}^{n}\left(\frac{1}{\sqrt{2πσ^2}}exp\left(\frac{-(x_i-𝜇)^2}{2σ^2}\right)\right)\\
\end{align*}$$

Aplicamos el logaritmo de ambos lados:

$$\begin{align*}log(L(𝜇,σ^2))&=log\left(∏_{i=1}^{n}\frac{1}{\sqrt{2πσ^2}}\cdot exp\left(\frac{-(x_i-𝜇)^2}{2σ^2}\right)\right)\\
&\hspace{1cm}\\
&=\sum_{i=1}^{n}\left[log\left(\frac{1}{\sqrt{2πσ^2}}\right)-\left(\frac{(x_i-𝜇)^2}{2σ^2}\right)\right]\\
&\hspace{1cm}\\
&=\sum_{i=1}^{n}\left[log(2πσ^2)^{-\frac{1}{2}}-\frac{(x_i-𝜇)^2}{2σ^2}\right]\\
&\hspace{1cm}\\
&=\sum_{i=1}^{n}\left[-\frac{1}{2}log(2πσ^2)-\frac{(x_i-𝜇)^2}{2σ^2}\right]\\
&\hspace{1cm}\\
&=\sum_{i=1}^{n}-\frac{1}{2}log(2πσ^2)-\sum_{i=1}^{n}\frac{(x_i-𝜇)^2}{2σ^2}\\
&\hspace{1cm}\\
&=-\frac{n}{2}log(2πσ^2)-\frac{1}{2σ^2}⋅\sum_{i=1}^{n}(x_i-𝜇)^2
\end{align*}$$

Derivaremos con respecto a $𝜇$:

$$\begin{align*}\frac{∂}{∂𝜇}log(L(𝜇,σ^2))&=\frac{∂}{∂𝜇}\left(-\frac{n}{2}log(2πσ^2)\right)-\frac{∂}{∂𝜇}\frac{1}{2σ^2}\sum_{i=1}^{n}(x_i-𝜇)^2\\
&\hspace{1cm}\\
&=-\frac{∂}{∂𝜇}\left(\frac{1}{2σ^2}\right)\sum_{i=1}^{n}(x_i-𝜇)^2\\
&\hspace{1cm}\\
&=-\frac{1}{2σ^2}\sum_{i=1}^{n}2(x_i-𝜇)(-1)\\
&\hspace{1cm}\\
&=\frac{1}{σ^2}\sum_{i=1}^{n}(x_i-𝜇)
\end{align*}$$

Igualamos a $0$ y despejamos $𝜇$:

$$\begin{align*}\frac{1}{σ^2}\sum_{i=1}^{n}(x_i-𝜇)&=0\\
&\hspace{1cm}\\
\sum_{i=1}^{n}(x_i-𝜇)&=σ^2(0)\\
&\hspace{1cm}\\
\sum_{i=1}^{n}x_i&=\sum_{i=1}^{n}𝜇\\
&\hspace{1cm}\\
n𝜇&=\sum_{i=1}^{n}x_i\\
&\hspace{1cm}\\
𝜇&=\frac{1}{n}\sum_{i=1}^{n}x_i=\bar{X}
\end{align*}$$

Ahora para encontrar el de $σ^2$, derivamos respecto a $σ^2$ al logaritmo de la función de verosimilitud:

$$\begin{align*}\frac{∂}{∂σ^2}log(L(𝜇,σ^2))&=\frac{∂}{∂σ^2}\left[-\frac{n}{2}log(2πσ^2)-\frac{1}{2σ^2}\sum_{i=1}^{n}(x_i-𝜇)^2\right]\\
&\hspace{1cm}\\
&=-\frac{n}{2}\frac{∂}{∂σ^2}log(2πσ^2)-\frac{∂}{∂σ^2}\left(\frac{1}{2σ^2}\sum_{i=1}^{n}(x_i-𝜇)^2\right)\\
&\hspace{1cm}\\
&=-\frac{n}{2}⋅\frac{2π}{2πσ^2}-\frac{∂}{∂σ^2}\frac{1}{2σ^2}\sum_{i=1}^{n}(x_1-𝜇)^2\\
&\hspace{1cm}\\
&=-\frac{n}{2σ^2}+\frac{1}{σ^4}\sum_{i=1}^{n}(x_i-𝜇)^2
\end{align*}$$

Igualamos a $0$ y despejamos $σ^2$:

$$\begin{align*} -\frac{n}{2σ^2}+\frac{1}{σ^4}\sum_{i=1}^{n}(x_i-𝜇)^2&=0\\
&\hspace{1cm}\\
2σ^4\left(\frac{-n}{2σ^2}+\frac{1}{σ^4}\sum_{i=1}^{n}(x_i-𝜇)^2\right)&=2σ^4(0)\\
&\hspace{1cm}\\
-σ^2n+2\sum_{i=1}^{n}(x_i-𝜇)^2&=0\\
&\hspace{1cm}\\
nσ^2&=2\sum_{i=1}^{n}(x_i-𝜇)^2\\
&\hspace{1cm}\\
σ^2&=\frac{2}{n}\sum_{i=1}^{n}(x_i-𝜇)^2
\end{align*}$$

Por lo tanto los EMV de $𝜇$ y $σ^2$ son:

$$\begin{align*}
\hat{𝜇}&=\bar{X}\\
&\hspace{1cm}\\
\hat{σ^2}&=\frac{2}{n}\sum_{i=1}^{n}(x_i-𝜇)^2
\end{align*}$$

## #Teorema de Fisher-Neyman

Si $X_1,\ldots,X_n$ es una m.a. con función de densidad $f(x;θ)$, y sea $T=t(x_1,\ldots,x_n)$ es una estadpistica cuya función de probabilidad $g(t;θ)$. Entonces $T$ es suficiente para $θ$ si y solo si:

$$L(x_1,\ldots,x_n;θ)=H(x_1,\ldots,x_n)\cdot g(t(x_1,\ldots,x_n);θ) \hspace{0.5cm} \text{, con } H(x_1,\ldots,x_n) \text{ que no dependa de }θ$$

A los estimadores que son insesgados de varianza uniformemente mínima se llaman UMVUE.

Sea $X_1,\ldots,X_n$ una m.a. de $f(\cdot;\theta)$ y $T*=t(x_1,\ldots,x_n)$ un estimador de $τ(θ)$ se define como un UMVUE de $τ(θ)$ si y solo si:

  1. $\mathbb{E}[T*]=τ(θ)$ es insesgado.
  2. $Var(T*)\leq Var(T)$ para cualquier otro estimador insesgaro $T=t(x_1,\ldots,x_n)$ de $τ(θ)$.

De manera similar, la desigualdar de Crámer-Rao establece un límite inferior para la varianza de cualquier estimador, de la siguiente manera:
$$Var(T)\geq \frac{\left[τ'(θ)\right]^2}{n\mathbb{E}\left[\left(\frac{∂ln(f(x;θ))}{∂θ}\right)^2\right]}$$

En caso de que la $Var(T)$ alcance el límite inferior, se dice que es un estimador insesgado óptimo.

# #Intervalos de Confianza

Un intervalo de confianza se define de la siguiente manera:

> <h3> <font color= "3672ec";>Sea $α ∈(0,1)$. Un intervalo de confianza para un parámetro desconocido $θ$ de una distribución de probabilidad es un intervalo aleatorio de la forma $(\hat{θ_1},\hat{θ_2})$, en donde $\hat{θ_1}$ y $\hat{θ_2}$ son estadísticas (funciones de una muestra aleatoria) tales que: $$\mathbb{P}(\hat{θ_1}<θ<\hat{θ_2})=1-α=γ$$</h3>

Donde $\hat{θ_1}$ y $\hat{θ_2}$ son estadísticos de $X_1,\ldots,X_n$. Y al intervalo aleatorio $(\hat{θ_1},\hat{θ_2})$ se le llama intervalo de $γ×100%$ de confianza.

Algunos intervalos de confianza conocidos son:

- Intervalo de confianza para la media $𝜇$ de una distribución normal $N(𝜇,σ^2)$, con varianza conocida:

$$I=\left[\bar{X}±Z_{\frac{α}{2}}\frac{σ}{\sqrt{n}}\right]$$

- Intervalo de confianza para la variación $σ^2$ de una distribución normal con $𝜇$ desconocida:

$$I=\left[±\frac{(n-1)s^2}{χ_{1-\frac{α}{2},n-1}^2}\right]$$

- Intervalo de confianza para $p$ de una distribución binomial de parametros $n,p,B(n,p)$:

$$I=\left[p_0±Z_{\frac{α}{2}}\sqrt{\frac{p_0(1-p_0)}{n}}\right]$$

- Intervalo de confianza para el parámetro $λ$ de una distribución Poisson:

$$I=\left[λ±Z_{\frac{α}{2}} \sqrt{\frac{λ}{2}}\right]$$


Para el intervalo de confianza para las diferencias de las medias con $X_1,\ldots,X_n$ una m.a. que se distriye como $N(𝜇_A,σ^2)$ y otra m.a. $Y_1,\ldots,Y_m$ que se distribuye como $N(𝜇_B,σ^2)$

Entonces: $\bar{X}∼N\left(𝜇_A,\frac{σ^2}{n}\right)$ y $\bar{Y}∼\left(𝜇_B,\frac{σ^2}{m}\right)$, con  $σ$ conocida:

$$(\bar{X}-\bar{Y})∼N\left(𝜇_A-𝜇_B,σ\left(\frac{1}{n}+\frac{1}{m}\right)\right)$$

Entonces el intervalo de confianza para $𝜇_A - 𝜇_B$ es:

$$I=\left(\left(\bar{X}-\bar{Y}\right)±Z_{1+\frac{α}{2}}σ\sqrt{\frac{1}{n}+\frac{1}{m}}\right)$$


$\hspace{1cm}$

Sea $X_1,\ldots,X_n$ una m.a. $X_i\sim N(𝜇,σ^2)$, el error estándar (SE) se define como la variabilidad de las medias muestrales con respecto a la desviación estandar y se ve de la siguiente forma:

$$SE=\frac{σ}{\sqrt{n}}$$

si la desviación estandar es $\sigma$ y el tamaño de la muestra es $n$.

En caso de que no se conozca $σ$ se utiliza la desviación estándar muestral $s$ y el SE se ve de la siguiente forma:

$$SE=\frac{s}{\sqrt{n}}$$

Una característica del error estándar es que en cuanto mayor sea $n$ es menor SE.

####<font color= "8f2bed"> Ejemplo:</font>
Para tratar de estimar la media de consumo por cliente en un gran restaurante, se reunieron datos de una muestra de 49 clientes durante 3 semanas.

**a)** Supongamos que la **desviación estándar de la población** es de $\$2.50$. ¿Cuál es el error estándar de la media?

**b)** Con un nivel de confianza del $95\%$, ¿cuál es el margen de error?

**c)** Si la **media de la muestra** es de $\$22.60$, cuál es el intervalo de confianza del $95\%$ para la media de la población?

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st
# Datos
n=49
sigma = 2.50
media_muestral = 22.60
confianza = 0.95

# a) Error estándar de la media
error_estandar = sigma / np.sqrt(n)
#error_estandar
print(f" a) Error estándar de la media: {error_estandar:.2f}")

# b) Margen de error
z = st.norm.ppf(1-(1-confianza) / 2)
margen_error = z * error_estandar
#margen_error
print(f" b) Margen de error con 95% de confianza: {margen_error:.2f}")

# c)
limite_inferior = media_muestral - margen_error
limite_superior = media_muestral + margen_error
print(f" c) Intervalo de confianza del 95%: ({limite_inferior:.2f},{limite_superior:.2f})")
```

####<font color= "8f2bed"> Ejemplo:</font>
El artículo *"Evaluation of a Ventilation Strategy to Prevent Barotrauma in Patients at High Risk for Acute Respiratory Distress Syndrome"* reportó sobre un experimento con 120 pacientes con anestesistas en cuidados intensivos (UCI), los cuales fueron divididos al azar en dos grupos, donde cada uno esta compuesto por 60 pacientes.

- Grupo A: promedio de permanencia = 14.1 horas
- Grupo B: promedio de permanencia = 17.5 horas
- Desviación estándar en ambos = 5.1 hrs

Encontrar un intervalo del $95\%$ de confianza para la diferecia de medias poblacionales: $(\mu_A - \mu_B)$  

```python
#Datos
n1 = 60
n2 = 60
media1 = 14.1
media2 = 17.5
sigma = 5.1
confianza = 0.95
alpha = 1-confianza
#gl = n1 + n2 -2

#valor critico
z = st.norm.ppf(1 - alpha / 2)

#Error estandar
error_est = sigma * np.sqrt(1/n1 + 1/n2)
dif_medias = media1 - media2
margen_error = z *error_est

lim_inf = dif_medias - margen_error
lim_sup = dif_medias + margen_error
print(f"Diferencia de las medias: {dif_medias:.2f}")
print(f"Intervalo de confianza del 95%: ({lim_inf:.2f},{lim_sup:.2f})")
```

####<font color= "f24421"> Ejercicio:</font>
En una encuesta de una Universidad, 225 estudiantes son seleccionados al azar de los que toman cálculo y se les pregunta si cálculo es su materia favorita. 100 estudiantes responden que el cálculo es su materia favorita. Proporcione un intervalo de confianza del 95 %  para la proporción de todos los estudiantes que toman cálculo y que la consideran su materia favorita.  

```python
import scipy.stats as st

# Datos
n = 225  # Tamaño de la muestra
x = 100  # Número de estudiantes que prefieren cálculo
p_muestra = x / n  # Proporción muestral
confianza = 0.95  # Nivel de confianza

# Calcular el intervalo de confianza
z = st.norm.ppf(1 - (1 - confianza) / 2)
margen_error = z * (p_muestra * (1 - p_muestra) / n)**0.5
limite_inferior = p_muestra - margen_error
limite_superior = p_muestra + margen_error

# Imprimir el resultado
print(f"Intervalo de confianza del 95% para la proporción: ({limite_inferior:.4f}, {limite_superior:.4f})")
```
